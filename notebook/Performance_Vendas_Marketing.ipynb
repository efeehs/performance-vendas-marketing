# Dimensão de Data

```python
dates = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="D"
)

dim_data = pd.DataFrame({
    "Date": dates
})

dim_data["Year"] = dim_data["Date"].dt.year

dim_data["Quarter"] = (
    "Q" + dim_data["Date"].dt.quarter.astype(str)
)

dim_data["Month_Number"] = dim_data["Date"].dt.month

dim_data["Month"] = dim_data["Date"].dt.month_name()

dim_data["Month_Short"] = (
    dim_data["Date"].dt.strftime("%b")
)

dim_data["Year_Month"] = (
    dim_data["Date"].dt.strftime("%Y-%m")
)

dim_data["Week"] = (
    dim_data["Date"].dt.isocalendar().week.astype(int)
)

dim_data["Day"] = dim_data["Date"].dt.day

dim_data["Day_of_Week"] = (
    dim_data["Date"].dt.day_name()
)

dim_data["Is_Weekend"] = (
    dim_data["Date"].dt.dayofweek >= 5
)

dim_data.head()
```

```python
print("Quantidade de registros:", len(dim_data))

print("\nPrimeiras linhas:")
display(dim_data.head())

print("\nÚltimas linhas:")
display(dim_data.tail())

print("\nTipos de dados:")
print(dim_data.dtypes)
```

```python
assert dim_data["Date"].min() == pd.Timestamp(START_DATE)
assert dim_data["Date"].max() == pd.Timestamp(END_DATE)
assert dim_data["Date"].is_unique
assert dim_data["Date"].notna().all()

print("Validação da dim_data: OK")
```

```python
volume_map = {
    "Wine": 750,
    "Sparkling Wine": 750,
    "Whisky": 750,
    "Gin": 700,
    "Vodka": 750,
    "Rum": 750,
    "Premium Beer": 500
}
```

```python
base_prices = []
base_costs = []
volumes = []

for _, product in dim_produto.iterrows():

    category = product["Category"]
    premium_level = product["Premium_Level"]

    min_price, max_price = price_ranges[
        category
    ][premium_level]

    base_price = np.random.uniform(
        min_price,
        max_price
    )

    margin_ranges = {
        "Standard": (0.18, 0.25),
        "Premium": (0.25, 0.35),
        "Super Premium": (0.30, 0.42)
    }

    min_margin, max_margin = margin_ranges[
        premium_level
    ]

    expected_margin = np.random.uniform(
        min_margin,
        max_margin
    )

    base_cost = base_price * (
        1 - expected_margin
    )

    volume = volume_map[category]

    base_prices.append(round(base_price, 2))
    base_costs.append(round(base_cost, 2))
    volumes.append(volume)


dim_produto["Base_Price"] = base_prices
dim_produto["Base_Cost"] = base_costs
dim_produto["Volume_ml"] = volumes
```

```python
segment_weights = {
    "Supermarket": 0.25,
    "Restaurant": 0.20,
    "Bar": 0.18,
    "Hotel": 0.10,
    "Specialty Store": 0.17,
    "E-commerce": 0.10
}
```

```python
customer_type_weights = {
    "Key Account": 0.08,
    "Strategic": 0.22,
    "Standard": 0.70
}
```

```python
region_weights = {
    "Southeast": 0.52,
    "South": 0.18,
    "Northeast": 0.14,
    "Central-West": 0.10,
    "North": 0.06
}
```

```python
NUM_CUSTOMERS = 250

customers = []

states = list(locations.keys())

state_pool = []

for state, info in locations.items():
    region = info["Region"]

    state_pool.extend(
        [state] * max(1, int(region_weights[region] * 100))
    )


for i in range(1, NUM_CUSTOMERS + 1):

    state = random.choice(state_pool)

    region = locations[state]["Region"]

    city = random.choice(
        locations[state]["Cities"]
    )

    segment = random.choices(
        list(segment_weights.keys()),
        weights=list(segment_weights.values()),
        k=1
    )[0]

    customer_type = random.choices(
        list(customer_type_weights.keys()),
        weights=list(customer_type_weights.values()),
        k=1
    )[0]

    customer = {
        "Customer_ID": f"C{i:04d}",
        "Customer": f"Cliente {i:03d}",
        "Segment": segment,
        "Customer_Type": customer_type,
        "State": state,
        "City": city,
        "Region": region
    }

    customers.append(customer)


dim_cliente = pd.DataFrame(customers)
```

# Fato de Vendas

```python
sales_channel_weights = {
    "SC01": 0.35,  # Retail
    "SC02": 0.30,  # Food Service
    "SC03": 0.20,  # E-commerce
    "SC04": 0.15   # Distributor
}
```

```python
segment_channel_weights = {
    "Supermarket": {
        "SC01": 0.65,
        "SC02": 0.10,
        "SC03": 0.05,
        "SC04": 0.20
    },

    "Restaurant": {
        "SC01": 0.05,
        "SC02": 0.75,
        "SC03": 0.05,
        "SC04": 0.15
    },

    "Bar": {
        "SC01": 0.10,
        "SC02": 0.70,
        "SC03": 0.05,
        "SC04": 0.15
    },

    "Hotel": {
        "SC01": 0.10,
        "SC02": 0.65,
        "SC03": 0.05,
        "SC04": 0.20
    },

    "Specialty Store": {
        "SC01": 0.45,
        "SC02": 0.10,
        "SC03": 0.25,
        "SC04": 0.20
    },

    "E-commerce": {
        "SC01": 0.05,
        "SC02": 0.02,
        "SC03": 0.90,
        "SC04": 0.03
    }
}
```

```python
seasonality = {
    1: 1.15,
    2: 1.05,
    3: 0.95,
    4: 0.90,
    5: 0.95,
    6: 1.05,
    7: 1.10,
    8: 1.00,
    9: 0.95,
    10: 1.00,
    11: 1.20,
    12: 1.45
}
```

```python
customer_type_factor = {
    "Standard": 1.0,
    "Strategic": 2.0,
    "Key Account": 4.0
}
```

```python
premium_factor = {
    "Standard": 1.00,
    "Premium": 0.75,
    "Super Premium": 0.45
}
```

```python
NUM_SALES = 60000

sales_records = []

available_dates = dim_data["Date"].tolist()

customer_weights = []

for _, customer in dim_cliente.iterrows():

    customer_weights.append(
        customer_type_factor[
            customer["Customer_Type"]
        ]
    )


for i in range(1, NUM_SALES + 1):

    date = random.choice(available_dates)

    month = date.month

    seasonal_factor = seasonality[month]

    customer_index = random.choices(
        range(len(dim_cliente)),
        weights=customer_weights,
        k=1
    )[0]

    customer = dim_cliente.iloc[
        customer_index
    ]

    customer_id = customer["Customer_ID"]

    segment = customer["Segment"]


    product_index = random.randrange(
        len(dim_produto)
    )

    product = dim_produto.iloc[
        product_index
    ]

    product_id = product["Product_ID"]

    premium_level = product["Premium_Level"]

    base_price = product["Base_Price"]

    base_cost = product["Base_Cost"]


    channel_ids = list(
        segment_channel_weights[
            segment
        ].keys()
    )

    channel_weights = list(
        segment_channel_weights[
            segment
        ].values()
    )

    channel_id = random.choices(
        channel_ids,
        weights=channel_weights,
        k=1
    )[0]



    base_quantity = random.randint(
        1,
        12
    )

    quantity = (
        base_quantity
        * seasonal_factor
        * premium_factor[premium_level]
        * customer_type_factor[
            customer["Customer_Type"]
        ]
    )

    quantity = max(
        1,
        int(round(quantity))
    )



    discount = np.random.choice(
        [
            0,
            0.03,
            0.05,
            0.08,
            0.10,
            0.15
        ],
        p=[
            0.30,
            0.15,
            0.20,
            0.15,
            0.15,
            0.05
        ]
    )



    unit_price = (
        base_price
        * np.random.uniform(
            0.95,
            1.05
        )
    )

    unit_price = round(
        unit_price * (1 - discount),
        2
    )



    revenue = (
        quantity
        * unit_price
    )



    cost = (
        quantity
        * base_cost
    )


    active_campaigns = dim_campanha[
        (dim_campanha["Start_Date"] <= date) &
        (dim_campanha["End_Date"] >= date)
    ]

    campaign_id = None

    if len(active_campaigns) > 0:


        if random.random() < 0.35:

            campaign = active_campaigns.sample(
                1
            ).iloc[0]

            campaign_id = campaign[
                "Campaign_ID"
            ]


    sales_records.append({
        "Sale_ID": f"S{i:06d}",
        "Date": date,
        "Product_ID": product_id,
        "Customer_ID": customer_id,
        "Channel_ID": channel_id,
        "Campaign_ID": campaign_id,
        "Quantity": quantity,
        "Unit_Price": unit_price,
        "Discount": discount,
        "Revenue": round(revenue, 2),
        "Cost": round(cost, 2)
    })


fato_vendas = pd.DataFrame(
    sales_records
)
```

# Fato de Marketing

```python
marketing_parameters = {
    "MC01": {  # Instagram
        "cpm": 18,
        "ctr": 0.012,
        "engagement_rate": 0.045,
        "lead_rate": 0.025,
        "conversion_rate": 0.12
    },

    "MC02": {  # Google Ads
        "cpm": 35,
        "ctr": 0.028,
        "engagement_rate": 0.018,
        "lead_rate": 0.055,
        "conversion_rate": 0.18
    },

    "MC03": {  # YouTube
        "cpm": 22,
        "ctr": 0.009,
        "engagement_rate": 0.030,
        "lead_rate": 0.020,
        "conversion_rate": 0.10
    },

    "MC04": {  # Influencers
        "cpm": 28,
        "ctr": 0.016,
        "engagement_rate": 0.065,
        "lead_rate": 0.035,
        "conversion_rate": 0.14
    },

    "MC05": {  # Events
        "cpm": 75,
        "ctr": 0.004,
        "engagement_rate": 0.080,
        "lead_rate": 0.045,
        "conversion_rate": 0.20
    },

    "MC06": {  # Email Marketing
        "cpm": 8,
        "ctr": 0.045,
        "engagement_rate": 0.025,
        "lead_rate": 0.080,
        "conversion_rate": 0.24
    }
}
```

```python
marketing_records = []

for _, campaign in dim_campaign.iterrows():

    campaign_id = campaign["Campaign_ID"]
    channel_id = campaign["Marketing_Channel_ID"]

    start_date = campaign["Start_Date"]
    end_date = campaign["End_Date"]
    budget = campaign["Budget"]

    params = marketing_parameters[channel_id]

    campaign_dates = pd.date_range(
        start=start_date,
        end=end_date,
        freq="D"
    )

    number_of_days = len(campaign_dates)

    daily_budget = budget / number_of_days

    for date in campaign_dates:

        spend = daily_budget * np.random.uniform(
            0.75,
            1.25
        )

        impressions = (
            spend / params["cpm"]
        ) * 1000

        clicks = (
            impressions *
            params["ctr"] *
            np.random.uniform(0.85, 1.15)
        )

        engagements = (
            impressions *
            params["engagement_rate"] *
            np.random.uniform(0.85, 1.15)
        )

        leads = (
            clicks *
            params["lead_rate"] *
            np.random.uniform(0.80, 1.20)
        )

        conversions = (
            leads *
            params["conversion_rate"] *
            np.random.uniform(0.80, 1.20)
        )

        marketing_records.append({
            "Date": date,
            "Campaign_ID": campaign_id,
            "Marketing_Channel_ID": channel_id,
            "Spend": round(spend, 2),
            "Impressions": int(impressions),
            "Clicks": int(clicks),
            "Engagements": int(engagements),
            "Leads": int(leads),
            "Conversions": int(conversions)
        })


fato_marketing = pd.DataFrame(
    marketing_records
)
```
